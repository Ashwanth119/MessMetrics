```python
from transformers import pipeline

classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

text = "abs and 6 pack"

labels = [
    # "feeling sad or emotionally low",
    "tired or low energy",
    "feeling hungry",
    "wanting to lose weight",
    "wanting to gain muscle",
    "feeling energetic"
]

result = classifier(text, candidate_labels=labels)

print(result["labels"])
print(result["scores"])
```

```python
result=classifier(""I'm feeling tired and have no energy")
```
