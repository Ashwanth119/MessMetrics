```python

import pandas as pd
import os

a = []
columns = ['hostel_name', 'day', 'item_name', 'meal']

for i in data[1:]:
    hostel_name = i['name']

    for j in i['mess']:
        day = j['day']

        for meal in ['breakfast', 'lunch', 'snacks', 'dinner']:
            items = j[meal].split(',')

            for item in items:
                item = item.strip()

                if item:
                    a.append([
                        hostel_name,
                        day,
                        item,
                        meal
                    ])

df = pd.DataFrame(a, columns=columns)

folder = os.path.join(os.getcwd(), 'mess_menus')
os.makedirs(folder, exist_ok=True)

df.to_csv(
    os.path.join(folder, 'all_hostel_menus.csv'),
    index=False
)

print("All hostel menus saved!")
print(df.head())
```

```python
def calculate_calories(age, height, weight, sex, activity):
    # Calculate BMR
    if sex.lower() == "male":
        bmr = 10 * weight + 6.25 * height - 5 * age + 5
    elif sex.lower() == "female":
        bmr = 10 * weight + 6.25 * height - 5 * age - 161
    else:
        raise ValueError("sex must be 'male' or 'female'")

    # Calculate TDEE
    activity_factors = {
        "sedentary": 1.2,
        "light": 1.375,
        "moderate": 1.55,
        "very_active": 1.725,
        "extra_active": 1.9
    }

    if activity not in activity_factors:
        raise ValueError("Invalid activity level")

    tdee = bmr * activity_factors[activity]

    return {
        "bmr": round(bmr),
        "maintenance_calories": round(tdee)
    }


result = calculate_calories(
    age=22,
    height=175,
    weight=75,
    sex="male",
    activity="moderate"
)

print(result)
```
